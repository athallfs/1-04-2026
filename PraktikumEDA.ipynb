**LATIHAN 1**

```python
import numpy as np

harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Jika menggunakan NumPy array, harga * 0.9 akan mengalikan setiap elemen dengan 0,9 sehingga semua harga otomatis mendapat diskon 10%. Sedangkan [5000,7000,3000] * 0.9 pada Python list akan menghasilkan TypeError karena list tidak mendukung perkalian dengan bilangan desimal. Pada list, operator * digunakan untuk mengulang isi list dengan bilangan bulat. NumPy array lebih cocok untuk operasi perhitungan pada banyak data karena mendukung operasi secara langsung pada setiap elemen.

**LATIHAN 2**

```python
import pandas as pd
data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Kolom yang memiliki data kosong adalah menu dan terjual. Data kosong pada kolom menu ditampilkan sebagai None, sedangkan pada kolom terjual ditampilkan sebagai NaN. Jika data langsung dianalisis tanpa dibersihkan, hasil perhitungan seperti total, rata-rata, atau jumlah penjualan dapat menjadi tidak akurat karena ada data yang hilang. Oleh karena itu, data kosong perlu diperiksa dan ditangani terlebih dahulu.

**LATIHAN 3**


```python
df = pd.read_csv('data_kantin.csv')

print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total adalah menu dan terjual. Kolom menu memiliki 107 data terisi dari 110 baris, sehingga terdapat 3 data kosong. Kolom terjual memiliki 104 data terisi, sehingga terdapat 6 data kosong. Artinya, kedua kolom tersebut memiliki missing value (data yang hilang) yang perlu dibersihkan atau ditangani sebelum dilakukan analisis lebih lanjut.

**LATIHAN 4**

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Kolom terjual diisi dengan 0 karena data yang kosong masih bisa dianggap sebagai tidak ada penjualan yang tercatat, sehingga data tetap dapat digunakan untuk perhitungan. Sedangkan baris dengan menu kosong dihapus karena nama menu merupakan informasi penting untuk mengetahui barang apa yang dijual. Jika nama menu tidak diketahui, data tersebut sulit digunakan dan dapat membuat analisis menjadi tidak jelas.

**LATIHAN 5**

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Setelah drop_duplicates(), jumlah baris dataset menjadi lebih sedikit atau tetap sama dibanding sebelumnya, tergantung apakah terdapat baris yang duplikat. Baris yang sama persis akan dihapus sehingga tidak dihitung berulang. Memastikan tipe data (dtypes) benar penting agar proses analisis dan perhitungan berjalan dengan tepat. Misalnya, kolom harga harus bertipe integer agar dapat digunakan untuk operasi matematika seperti menghitung total atau rata-rata harga tanpa masalah.

**LATIHAN 6**

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total pendapatan tertinggi adalah Nasi Goreng, dengan total pendapatan sebesar Rp3.924.000.

Informasi ini dapat membantu kantin sekolah menentukan menu yang perlu diprioritaskan, menjaga ketersediaan stok bahan baku, dan mengatur strategi penjualan agar pendapatan kantin meningkat.

Sebagai langkah berikutnya, kantin juga bisa membandingkan jumlah porsi terjual untuk mengetahui apakah pendapatan tinggi disebabkan oleh banyaknya penjualan atau harga per porsi.
